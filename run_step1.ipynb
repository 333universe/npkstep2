## Try your own claims

```python
from npkmath.engine import NPKCoreEngine
from npkmath.signals import AIFeedbackInterface
from npkmath.pipeline import verify_equality_and_emit, verify_matrix_and_emit

engine, fb = NPKCoreEngine(), AIFeedbackInterface()

verify_equality_and_emit(engine, fb, 'my-claim-1', 'sin(x)**2 + cos(x)**2', '1')
verify_equality_and_emit(engine, fb, 'my-claim-2', '(x+y)**2', 'x**2 + y**2')
verify_matrix_and_emit(engine, fb, 'my-matrix-1', 'orthogonal', [['1','2'],['3','4']])

print(fb.export_for_prompt())
print(fb.get_summary())
```
